```python
from datetime import datetime
import math

def hitung_durasi(waktu_keluar, waktu_masuk):
    durasi = waktu_keluar - waktu_masuk
    durasi_jam = math.ceil(durasi.total_seconds() / 3600)
    return durasi_jam
```

```python
from datetime import datetime, timedelta
import math


TARIF = {
    "mobil": 5000,
    "motor": 2000
}



riwayat = [

    {"plat": "M01", "jenis": "motor", "masuk": "2026-01-15 08:00:00", "keluar": "2026-01-15 10:00:00"},
    {"plat": "M02", "jenis": "motor", "masuk": "2026-01-15 09:00:00", "keluar": "2026-01-15 11:00:00"},
    {"plat": "C01", "jenis": "mobil", "masuk": "2026-01-15 08:00:00", "keluar": "2026-01-15 10:00:00"},
    {"plat": "C02", "jenis": "mobil", "masuk": "2026-01-15 09:00:00", "keluar": "2026-01-15 12:00:00"},

    {"plat": "M03", "jenis": "motor", "masuk": "2026-01-14 10:00:00", "keluar": "2026-01-14 12:00:00"},
    {"plat": "M04", "jenis": "motor", "masuk": "2026-01-12 13:00:00", "keluar": "2026-01-12 15:00:00"},
    {"plat": "C03", "jenis": "mobil", "masuk": "2026-01-13 08:00:00", "keluar": "2026-01-13 11:00:00"},
    {"plat": "C04", "jenis": "mobil", "masuk": "2026-01-11 14:00:00", "keluar": "2026-01-11 16:00:00"},
    {"plat": "C05", "jenis": "mobil", "masuk": "2026-01-10 07:00:00", "keluar": "2026-01-10 09:00:00"},

    {"plat": "M05", "jenis": "motor", "masuk": "2026-01-05 09:00:00", "keluar": "2026-01-05 11:00:00"},
    {"plat": "M06", "jenis": "motor", "masuk": "2026-01-02 10:00:00", "keluar": "2026-01-02 12:00:00"},
    {"plat": "C06", "jenis": "mobil", "masuk": "2026-01-04 15:00:00", "keluar": "2026-01-04 18:00:00"},
    {"plat": "C07", "jenis": "mobil", "masuk": "2026-01-03 11:00:00", "keluar": "2026-01-03 14:00:00"},
    {"plat": "C08", "jenis": "mobil", "masuk": "2026-01-02 08:00:00", "keluar": "2026-01-02 10:00:00"},

    {"plat": "M07", "jenis": "motor", "masuk": "2025-12-25 09:00:00", "keluar": "2025-12-25 11:00:00"},
    {"plat": "M08", "jenis": "motor", "masuk": "2025-12-20 10:00:00", "keluar": "2025-12-20 12:00:00"},
    {"plat": "C09", "jenis": "mobil", "masuk": "2025-12-28 08:00:00", "keluar": "2025-12-28 11:00:00"},
    {"plat": "C10", "jenis": "mobil", "masuk": "2025-12-15 07:00:00", "keluar": "2025-12-15 10:00:00"},
    {"plat": "C11", "jenis": "mobil", "masuk": "2025-12-10 13:00:00", "keluar": "2025-12-10 16:00:00"},
    {"plat": "C12", "jenis": "mobil", "masuk": "2025-12-05 14:00:00", "keluar": "2025-12-05 17:00:00"},
]

def hitung_rekap(riwayat, mulai):
    rekap = {
        "motor": {"jumlah": 0, "durasi": 0, "pendapatan": 0},
        "mobil": {"jumlah": 0, "durasi": 0, "pendapatan": 0},
    }

    for r in riwayat:
        masuk = datetime.strptime(r["masuk"], "%Y-%m-%d %H:%M:%S")
        keluar = datetime.strptime(r["keluar"], "%Y-%m-%d %H:%M:%S")

        if masuk < mulai:
            continue

        jam = max(1, math.ceil((keluar - masuk).total_seconds() / 3600))
        jenis = r["jenis"]

        rekap[jenis]["jumlah"] += 1
        rekap[jenis]["durasi"] += jam
        rekap[jenis]["pendapatan"] += jam * TARIF[jenis]

    return rekap


def tampilkan(judul, rekap):
    print(f"\n=== {judul} ===")

    total_kendaraan = total_jam = total_uang = 0

    for jenis in ["motor", "mobil"]:
        data = rekap[jenis]
        print(
            f"{jenis.upper():<6} : "
            f"{data['jumlah']} kendaraan | "
            f"{data['durasi']} jam | "
            f"Rp {data['pendapatan']:,}".replace(",", ".")
        )

        total_kendaraan += data["jumlah"]
        total_jam += data["durasi"]
        total_uang += data["pendapatan"]

    print(
        f"TOTAL  : {total_kendaraan} kendaraan | "
        f"{total_jam} jam | "
        f"Rp {total_uang:,}".replace(",", ".")
    )


now = datetime(2026, 1, 15, 23, 59, 59)

hari_ini = datetime(now.year, now.month, now.day)
minggu_ini = now - timedelta(days=7)
bulan_ini = datetime(now.year, now.month, 1)
tahun_ini = datetime(now.year, 1, 1)

tampilkan("HARI INI", hitung_rekap(riwayat, hari_ini))
tampilkan("MINGGU INI", hitung_rekap(riwayat, minggu_ini))
tampilkan("BULAN INI", hitung_rekap(riwayat, bulan_ini))
tampilkan("TAHUN INI", hitung_rekap(riwayat, tahun_ini))
```
